```python
import numpy as np
```

```python
# 1. Convert grades: "Pass" if score >= 50, else "Fail"
scores = np.array([45, 67, 32, 90, 58])
result = np.where(scores>50, "Pass", "Fail")
result
```

```python
# 2. Make numbers positive: replace negatives with 0
nums = np.array([-4, -1, 3, -7, 8, 0])
replace = np.where(nums<0, 0, nums)
replace
```

```python
# 3. Apply discount: 10% off if price > 100
prices = np.array([80, 120, 150, 90, 200])
discount = np.where(prices>100, prices*.90, prices)
discount
```

```python
# 4. Categorize ages: "Child" if under 18, else "Adult"
ages = np.array([8, 15, 22, 35, 60, 70])
category = np.where(ages < 18, "Child", "Adult")
category
```

```python
# 5. Label numbers: "Even" if divisible by 2, else "Odd"
nums = np.array([1, 2, 3, 4, 5, 6])
labels = np.where(nums % 2 == 0, "Even", "Odd")
labels
```

```python
# 6. Cap values at a maximum of 100 (upper bound)
values = np.array([10, 25, 50, 75, 120])
capped = np.where(values > 100, 100, values)
capped
```

```python
# Q1. Weather report: label each day "Hot" if temperature > 35, else "Normal"
temperatures = np.array([28, 36, 41, 33, 30, 38, 25])
label = np.where(temperatures>35,"HOT","NORMAL")
label
```

```python
# Q2. Absent students are marked -1: replace -1 with 0 and keep other marks as they are
marks = np.array([78, -1, 56, 89, -1, 45])
absent = np.where(marks==-1, 0, marks)
absent
```

```python
# Q3. Free delivery: delivery charge is 0 if order amount >= 499, else 40
#     Then find the total amount to pay (order + delivery charge)
orders = np.array([250, 499, 799, 120, 1500])
delivery = np.where(orders>=499, orders, orders+40)
delivery
```

```python
# Q4. Salary hike: 20% hike if salary < 30000, else 10% hike. Find the new salaries
salaries = np.array([25000, 45000, 18000, 60000, 30000])
hike = np.where(salaries<30000,salaries*1.20, salaries*1.10)
hike
```

```python
# Q5. Phone battery status (use two arrays together):
#     "Charging"   -> if the phone is plugged in
#     "Charge Now" -> if NOT plugged in AND battery < 20
#     "OK"         -> otherwise
battery = np.array([15, 80, 5, 45, 19, 10])
is_charging = np.array([False, False, True, False, False, True])

status = np.where(is_charging == True,"Charging",
                  np.where(battery < 20, "Charge Now" , "Ok"))
status
```

```python
# Q6. (Challenge) Stock report using yesterday's and today's prices:
#     First find the % change = (today - yesterday) / yesterday * 100
#     Then label: "Big Jump" if change > 5, "Up" if change > 0,
#                 "Down" if change < 0, else "No Change"
#     Hint: put one np.where inside another
yesterday = np.array([200, 150, 320, 90, 500, 75])
today     = np.array([215, 150, 310, 93, 500, 72])
# Your code here
change = (today - yesterday) / yesterday * 100
change
```

```python
label = np.where(change>5, "Big Jump",
                np.where(change > 0,"Up",
                         np.where(change<0,"Down", "No change")))
label
```
