```python
import numpy as np
```

```python
ages = np.array([12, 25, 17, 34, 45, 8, 19, 62, 29, 15, 51, 38, 70, 22, 16, 41])
ages
```

```python
adult_mask = ages>=18
child_mask = ages<18
teens_mask = (ages>=15)&(ages<=18)
working_mask = (ages>=18)&(ages<=60)
special_mask = (ages<10)|(ages>60)

print(f"ages[adult_mask] = {ages[adult_mask]}")
print(f"ages[child_mask] = {ages[child_mask]}")
print(f"ages[teens_mask] = {ages[teens_mask]}")
print(f"ages[working_mask] = {ages[working_mask]}")
print(f"ages[special_mask] = {ages[special_mask]}")
```
