```python
import numpy as np
```

```python
scores = np.array([45,67,72,59,88,34])
```

CHANGE ONE VALUE

```python
scores[0] = 50
scores
```

REPLACE A SLICE

```python
scores[1:3] = [70, 75]
scores
```

UPDATE WITH A CONDITION


```python
scores[scores<40] = 40
scores
```

UPDATE SELECTED POSITIONS

```python
scores[[3,4]] = [60,90]
scores
```
